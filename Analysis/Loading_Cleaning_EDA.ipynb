**Workforce cut based on Sectors**

```python
# Grouping up data
sector_job_cuts = (
    df.groupby('sector')['jobs_cut']
      .sum()
      .sort_values(ascending=True)
)

fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

# Normalize values between 0 and 1
norm = mpl.colors.PowerNorm(
    gamma=0.5,
    vmin=sector_job_cuts.min(),
    vmax=sector_job_cuts.max()
)

# Blue gradient
cmap = mpl.cm.Blues

# Give each bar a color according to its value
colors = cmap(norm(sector_job_cuts.values))

# Plot bars
ax.barh(
    sector_job_cuts.index,
    sector_job_cuts.values,
    height=0.8,
    color=colors
)

# Smaller Y-axis labels
ax.tick_params(axis='both', labelsize=9,colors = 'white')


#  applying labels
for i,value in enumerate(sector_job_cuts.values):
    plt.text(value+100,i,value,va='center',size=10,color='white')

# Remove grid
ax.grid(False)

# Spines
for spine in ax.spines.values():
    spine.set_color('white')

ax.set_xlabel("Number of Jobs",color='white')
ax.set_title("Workforce Layoffs in Different Sectors",color='white')

plt.tight_layout()
plt.show()
```

```python
# Slicing Columns
ai_investment_job_cut = df[['company','simultaneous_ai_investment_bn','jobs_cut']].sort_values('simultaneous_ai_investment_bn')
top5_investors = ai_investment_job_cut.tail(5).index.to_list()


# Plotting Chart
fig, ax = plt.subplots(figsize=(8,6))
fig.patch.set_facecolor('black')
ax.set_facecolor('black')
sns.scatterplot(alpha=.7,sizes = (50,500),data=ai_investment_job_cut,x='simultaneous_ai_investment_bn',y='jobs_cut',size='jobs_cut',legend=False,color='#3730A3')

# Plotting labels
for i,index in enumerate(top5_investors):
    plt.text(ai_investment_job_cut.loc[index,'simultaneous_ai_investment_bn']+3,ai_investment_job_cut.loc[index,'jobs_cut'],ai_investment_job_cut.loc[index,'company'],size=9,color='white')


# Cleaning Graph
ax = plt.gca()
ax.yaxis.set_major_formatter(FuncFormatter(lambda x, pos: f'{x/1000:.0f}K'))
ax.tick_params(axis='y',which='major', left=True, right=False, length=6, width=1.2, colors='white')
ax.tick_params(axis='x',which='major', bottom=True, right=False, length=6, width=1.2, colors='white')
for spine in ax.spines.values():
    spine.set_color('white')
sns.despine()
plt.title('AI Investment vs Jobs Cut',size='13',color='white')
plt.ylabel('Number of Layoffs',size=9,color='white')
plt.xlabel('USD in Billions Invested by Companies',size=9,color='white')
plt.tight_layout()
plt.show()
```

**- Major tech companies making massive AI investments (over $40 billion) all show significant layoffs of 15,000 or more, though the highest spenders don't necessarily have the most layoffs.**

**- No clear relationship is observed between simultaneous AI investment and the number of jobs cut.**

**- Oracle stands out as an outlier, recording the highest number of layoffs (~30K) despite a mid-tier AI investment level (~$40B), indicating aggressive operational restructuring relative to its spending.**

**How many Jobs are Replaced by AI**

```python
reasons = df['reason_stated'].to_list()
```

```python
mapping = {}

for reason in reasons:
    if "AI" in reason:
        mapping[reason] = "AI"
    else:
        mapping[reason] = reason

mapping
```

```python
df['reason_AI'] = df['reason_stated'].apply(lambda x: mapping[x])
```

```python
df_cut_count = df.groupby('reason_AI')[['jobs_cut']].sum()
df_cut_count['total_jobs_cut'] = df.jobs_cut.sum()
df_cut_count['percentage_job_cut_reason'] = (df_cut_count['jobs_cut']/df_cut_count['total_jobs_cut']*100).round(2)
df_cut_count_p = df_cut_count.sort_values('percentage_job_cut_reason')
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor('black')
ax.set_facecolor('black')

ax.tick_params(axis='both', colors='white')

for spine in ax.spines.values():
    spine.set_color('white')

plt.bar(
    df_cut_count_p.index,
    df_cut_count_p['percentage_job_cut_reason'],
    color='#4338CA',
    edgecolor='none',
    linewidth=0
)



for i, value in enumerate(df_cut_count_p['percentage_job_cut_reason']):
    plt.text(
        i,
        value + 0.5,
        f'{value:.2f}%',
        ha='center',
        color='white'
    )

plt.xlabel('', color='white')
plt.ylabel('Percentage of Total Job Cuts', color='white')
plt.title('Distribution of Job Cuts by Reason', color='white')
plt.xticks(rotation=45,ha='right')

sns.despine()

plt.tight_layout()
plt.show()
```

**AI-related reasons account for 61.88% of total job cuts, significantly higher than any individual non-AI reason. The remaining cuts are mainly associated with bureaucracy reduction, restructuring, cost reduction, and market slowdown.**
